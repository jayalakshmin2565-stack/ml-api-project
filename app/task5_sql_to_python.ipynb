```python
conn.autocommit = True

cursor = conn.cursor()

cursor.execute("""
IF DB_ID('MLDatabase') IS NULL
BEGIN
    CREATE DATABASE MLDatabase
END
""")

print("MLDatabase created successfully!")
```

```python
cursor.execute("""
SELECT name
FROM sys.databases
WHERE name = 'MLDatabase'
""")

for row in cursor:
    print(row[0])
```

```python
conn.close()

conn = pyodbc.connect(
    "DRIVER={SQL Server};"
    r"SERVER=Z14-55N\SQLEXPRESS;"
    "DATABASE=MLDatabase;"
    "Trusted_Connection=yes;"
)

print("Connected to MLDatabase successfully!")
```

```python
cursor = conn.cursor()

cursor.execute("""
IF OBJECT_ID('Customers', 'U') IS NULL
BEGIN
    CREATE TABLE Customers (
        CustomerID INT,
        Age INT,
        Income FLOAT,
        PurchaseAmount FLOAT
    )
END
""")

conn.commit()

print("Customers table created successfully!")
```

```python
cursor.execute("DELETE FROM Customers")

data = [
    (1, 22, 25000, 1200),
    (2, 28, 35000, 1800),
    (3, 35, 50000, 2500),
    (4, 42, 65000, 3200),
    (5, 30, 45000, 2100)
]

cursor.executemany("""
INSERT INTO Customers
(CustomerID, Age, Income, PurchaseAmount)
VALUES (?, ?, ?, ?)
""", data)

conn.commit()

print("Sample data inserted successfully!")
```

```python
query = "SELECT * FROM Customers"

df = pd.read_sql(query, conn)

df
```
